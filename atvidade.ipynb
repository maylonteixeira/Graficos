```python
from dash import Dash, dcc, html
import pandas as pd
import plotly.express as px

# 1. Ler o dataset enviado
df = pd.read_csv('ecommerce_estatistica.csv')

# 2. Criar os gráficos interativos com Plotly
fig_hist = px.histogram(df, x='Preço', nbins=30, title='Distribuição de Preços dos Produtos')
fig_scatter = px.scatter(df, x='N_Avaliações', y='Preço', color='Gênero', title='Relação Avaliações x Preço', hover_data=['Título'])

df_preco_genero = df.groupby('Gênero')['Preço'].mean().reset_index()
fig_bar = px.bar(df_preco_genero, x='Gênero', y='Preço', color='Gênero', title='Preço Médio por Gênero')

df_temporada = df['Temporada'].value_counts().reset_index()
df_temporada.columns = ['Temporada', 'Quantidade']
fig_pie = px.pie(df_temporada, names='Temporada', values='Quantidade', title='Distribuição por Temporada', hole=0.3)

# 3. Inicializar a aplicação Dash
app = Dash(__name__)

# 4. Estrutura visual da página
app.layout = html.Div(style={'backgroundColor': '#f9f9f9', 'padding': '20px'}, children=[
    html.H1("Dashboard de Análise de E-commerce", style={'textAlign': 'center'}),

    html.Div([
        html.Div([dcc.Graph(figure=fig_hist)], style={'width': '48%', 'display': 'inline-block'}),
        html.Div([dcc.Graph(figure=fig_scatter)], style={'width': '48%', 'display': 'inline-block'}),
    ]),

    html.Div([
        html.Div([dcc.Graph(figure=fig_bar)], style={'width': '48%', 'display': 'inline-block'}),
        html.Div([dcc.Graph(figure=fig_pie)], style={'width': '48%', 'display': 'inline-block'}),
    ])
])

# 5. Executar a aplicação com suporte a links públicos no Colab
if __name__ == '__main__':
    app.run(debug=True, port=8050)
```
